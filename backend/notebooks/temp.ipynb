```python
import glob, numpy as np, rasterio

for pre in sorted(glob.glob(r"C:\sen1floods11\S1Pre\*_S1Pre.tif")):
    key = pre.split("\\")[-1].replace("_S1Pre.tif", "")
    post = glob.glob(rf"C:\sen1floods11\**\{key}_S1Hand.tif", recursive=True)[0]
    with rasterio.open(pre) as a, rasterio.open(post) as b:
        same_grid = a.transform == b.transform and a.shape == b.shape
        p, q = a.read(1), b.read(1)
    ok = np.isfinite(p) & np.isfinite(q) & (q > -50)
    corr = np.corrcoef(p[ok], q[ok])[0, 1]
    print(f"{key:16} grid {same_grid}  pre {np.nanmedian(p):6.1f} dB  "
          f"post {np.median(q[ok]):6.1f} dB  missing {np.isnan(p).mean():4.0%}  corr {corr:.2f}")
```
